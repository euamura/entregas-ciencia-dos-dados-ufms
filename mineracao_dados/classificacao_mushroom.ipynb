```python
# aplica one-hot encoding para transformar atributos categóricos em numéricos

from sklearn.preprocessing import OneHotEncoder
import numpy as np

# cria o codificador
encoder = OneHotEncoder(
    handle_unknown="ignore",   # categoria não vista no treino vira tudo 0, em vez de dar erro
    sparse_output=False        # devolve matriz comum
)

# aprende as categorias de cada coluna e transforma
X_codificado = encoder.fit_transform(X)

print(X.shape)               # antes
print(X_codificado.shape)    # depois

```

```python
from sklearn.model_selection import train_test_split

X_treino, X_teste, y_treino, y_teste = train_test_split(
    X, y,
    test_size=0.30,     # 30% para teste
    stratify=y,         # mantém a proporção e/p nos dois conjuntos
    random_state=42     # mesmo sorteio a cada execução
)

# conferência 1: tamanhos
print("Treino:", X_treino.shape)
print("Teste: ", X_teste.shape)

# conferência 2: proporção das classes
print("\nBase completa:")
print(y.value_counts(normalize=True))
print("\nTreino:")
print(y_treino.value_counts(normalize=True))
print("\nTeste:")
print(y_teste.value_counts(normalize=True))
```

## Pipeline com One-Hot e AdaBoost

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier

adaboost = make_pipeline(
    OneHotEncoder(handle_unknown="ignore"),
    AdaBoostClassifier(
        estimator=DecisionTreeClassifier(max_depth=1),  # stump: o modelo fraco
        n_estimators=50,      # 50 stumps em sequência
        learning_rate=1.0,    # peso de cada stump
        random_state=42
    )
)

adaboost.fit(X_treino, y_treino)
```

```python
modelo = adaboost.named_steps["adaboostclassifier"]
nomes = adaboost.named_steps["onehotencoder"].get_feature_names_out()

# pergunta feita por cada um dos 10 primeiros stumps
for i, stump in enumerate(modelo.estimators_[:10]):
    atributo = nomes[stump.tree_.feature[0]]
    peso = modelo.estimator_weights_[i]
    print(f"Stump {i+1}: pergunta sobre '{atributo}' | peso do voto: {peso:.2f}")
```

## Avaliar teste

```python
from sklearn.metrics import accuracy_score, recall_score, confusion_matrix, classification_report

y_pred = adaboost.predict(X_teste)

print("Acurácia:", accuracy_score(y_teste, y_pred))
print("Recall (venenoso):", recall_score(y_teste, y_pred, pos_label="p"))

print("\nMatriz de confusão (linhas = real, colunas = previsto):")
print(confusion_matrix(y_teste, y_pred, labels=["p", "e"]))

print("\n", classification_report(y_teste, y_pred))
```
